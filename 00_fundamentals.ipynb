## reshaping,viewing, stacking, sqeezing and unsqueezing of tensors

```python
# reshaping -> reshapes the input tensor to a defined shape and also make change in memory
# view -> reshapes the input tensor to a defined shape but does not make change in memory
# stacking -> stacking tensors on top of each other
# squeeize -> removes all the dimensions of size 1
# unsqueeze -> adds a dimension of size 1 at a specified position
# permute -> rearranges the dimensions of a tensor in a specified order
```

```python
x=torch.arange(1.,11.)
x,x.shape
```

```python
x_reshaped=x.reshape(2,5)
x_reshaped,x_reshaped.shape
```

```python
# changing the value in reshaped will not change the value in memory
x_reshaped[0][0]=100
x,x_reshaped  # here x_reshaped is not forced to copy thats why changing the value also changes the value in x
```

```python
# tracking back x
x[0]=1
x
```

```python
# changing the view
z=x.view(2,5)
z,z.shape
```

```python
# changing value in view will change the original tensor
z[0][0]=100
x,z
```

```python
# stacking
torch.stack([x,x,x],dim=0)
```

```python
torch.vstack([x,x,x])
```

```python
torch.stack([x,x,x],dim=1)
```

```python
torch.hstack([x,x,x,torch.arange(15,19)])
```

```python
# squeeze
x=torch.arange(1.,10.).reshape(1,3,3,1)
x
```

```python
torch.squeeze(torch.squeeze(x)),torch.squeeze(x).shape
```

```python
x.squeeze(),x.squeeze().shape
```

```python
y=x.squeeze()
y
```

```python
y=y.reshape(9,1)
y.squeeze(),y.squeeze().shape
```

```python
x=torch.arange(1.,10.).reshape(1,1,3,3,1)
x_squeezed=torch.squeeze(x)
x_squeezed,x_squeezed.shape #it removes all the dimensions of size 1
```

```python
x_unsqueeze=x_squeezed.unsqueeze(dim=0)
x_unsqueeze,x_unsqueeze.shape #it adds a dimension of size 1 at the specified position
```

```python
x_unsqueeze=x_squeezed.unsqueeze(dim=1)
x_unsqueeze,x_unsqueeze.shape #it adds a dimension of size 1 at the specified position
```

```python
x=torch.arange(1.,10.).reshape(1,3,3)
x,x.shape
```

```python
x.permute(2,0,1),x.permute(2,0,1).shape #it rearranges the dimensions of a tensor in a specified order
```

```python
# images
x_original=torch.rand(size=(224,224,3)) #(height,width,channels)
x_original.shape
x_image=torch.permute(x_original,(2,0,1)) # rearranging the dimensions to (channels,height,width) which is the common format for image tensors in deep learning

x_image.shape
```
